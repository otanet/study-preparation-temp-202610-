```python
# このセルはそのまま実行してください（1セル完結）
import ipywidgets as widgets
from IPython.display import display, Markdown, clear_output
import pandas as pd


# --- Q1〜Q5（選択式） ---
questions_data = [
    {
        "question": r"""
**Q1 画像データの処理とCPU・メモリの負荷に関する次の記述のうち、誤っているものを1つ選べ。**
(A) 大きな画像のコピーを繰り返すと、メモリへの読み書きが増え、処理時間に影響する場合がある。
(B) 画像を圧縮して保存サイズを小さくすれば、圧縮処理を含めたCPUの負荷も必ず小さくなる。
(C) カメラのHzが低下したときは、CPUを使うプロセスを調べ、圧縮処理などの負荷を確認する。
(D) 講義の体験談では、depth画像のPNG圧縮がCPUを逼迫させていると分かり、RVLへ変更した。
(E) 圧縮方式の変更後は、保存サイズだけでなく、CPU使用率や画像のHz、読出し側の対応も確認する。
""",
        "options": ["A", "B", "C", "D", "E"]
    },
    {
        "question": r"""
**Q2 ロボット内のPCと外部GPUマシンの時刻同期に関する次の記述のうち、誤っているものを1つ選べ。**
(A) 複数のPCでセンサ情報を処理する場合、時計のずれはタイムスタンプを使う処理に影響することがある。
(B) chronyによる同期を調べるときは、一方のPCだけでなく、両方のPCで同期先や状態を確認する。
(C) センサのタイムスタンプについては、計測時刻か受信時刻か、どの時計を使っているかを確認する。
(D) PC同士の時計が同期していれば、センサ固有の時計や撮影タイミングも自動的に一致する。
(E) TFを使う処理では、要求した時刻に対応する座標変換が利用できるかを確認することが重要である。
""",
        "options": ["A", "B", "C", "D", "E"]
    },
    {
        "question": r"""
**Q3 Wi-Fi経由の遠隔操作とデータ転送に関する次の記述のうち、誤っているものを1つ選べ。**
(A) 遠隔操作画面とクラウドでは送信先が異なるため、同じWi-Fi回線を使っても帯域の競合は起こらない。
(B) 講義の体験談では、デバッグ用rvizでの画像の追加購読が通信を逼迫させ、操作映像がカクついた。
(C) 外部PCで映像を一度受け取り、外部側で各画面へ配る中継方式は、ロボットからの重複送信を減らせる。
(D) ロボット内に保存したデータをクラウドへ送る際も、遠隔操作と共有する通信経路の負荷を確認する。
(E) アップロードの帯域制限を導入した後は、遠隔操作の遅延やカクつきが改善したかを同じ条件で比較する。
""",
        "options": ["A", "B", "C", "D", "E"]
    },
    {
        "question": r"""
**Q4 DockerとLinuxの仮想化に関する次の記述のうち、誤っているものを1つ選べ。**
(A) Linuxコンテナはホストのカーネルを共有するため、利用できるカーネル機能はホストに依存する。
(B) namespaceは、プロセス一覧やネットワークなど、プロセスから見える範囲を分ける仕組みである。
(C) cgroupsでメモリ上限が設定されていると、ホストに空きがあってもコンテナ内でメモリ不足が起こり得る。
(D) ホストでは使えるカメラがコンテナで使えない場合、デバイスの割り当てや実行ユーザーの権限を確認する。
(E) 同じDockerイメージを配れば、ホストのGPUドライバの違いに関係なく、GPUの動作は保証される。
""",
        "options": ["A", "B", "C", "D", "E"]
    },
    {
        "question": r"""
**Q5 Linuxでのロボットのデバイス管理に関する次の記述のうち、誤っているものを1つ選べ。**
(A) udevルールでは、機器を識別できるシリアル番号などの属性に基づき、固定名のリンクを作成できる。
(B) Permission deniedが出たときは、デバイスファイルの所有者・グループ・権限と実行ユーザーを照合する。
(C) 一度カメラを/dev/video0として認識すれば、再起動や抜き差しの後も同じカメラがその名前に割り当てられる。
(D) 複数のUSBカメラを使う場合、接続先が異なるポートでも、上流の帯域を共有している可能性がある。
(E) カーネル更新後にGPUが使えなくなった場合、ドライバのモジュール読み込みやカーネルログを確認する。
""",
        "options": ["A", "B", "C", "D", "E"]
    },
]

# Markdownの行末に半角スペース2つを自動付与（改行表示用）
for q in questions_data:
    q["question"] = '\n'.join([line.rstrip() + '  ' for line in q["question"].split('\n')])

# ウィジェット生成（Q1〜Q5）
widgets_list = []
for i, q in enumerate(questions_data):
    display(Markdown(q["question"]))
    rb = widgets.RadioButtons(
        options=q["options"],
        description="回答",
        style={'description_width': 'initial'}
    )
    widgets_list.append(rb)
    display(rb)

button_layout = widgets.Layout(width='200px', height='50px')
# --- 提出ボタン ---
submit_button = widgets.Button(description="✅ 回答をCSVに保存", layout=button_layout)
output = widgets.Output()

def on_submit_all(b):
    with output:
        clear_output()
        # Q1〜Q5の検証
        try:
            answers_mc = [rb.value for rb in widgets_list]  # 文字列で扱う
        except Exception:
            print("⚠️ Q1〜Q5に未回答があります。すべての選択問題に回答してください。")
            return
        if any(a is None for a in answers_mc):
            print("⚠️ Q1〜Q5に未回答があります。すべての選択問題に回答してください。")
            return

        # CSV生成：id=0..4 → Q1〜Q5
        labels = answers_mc
        df = pd.Series(labels, name="label")
        df.to_csv("submission.csv", index_label="id", encoding="utf-8")

        print("✅ 保存完了！ → 'submission.csv'")

submit_button.on_click(on_submit_all)
display(submit_button, output)
```
