**以下のセルを修正せずに実行し、表示される各設問にお答えください。**

````python
# このセルはそのまま実行してください（1セル完結）
import ipywidgets as widgets
from IPython.display import display, Markdown, clear_output
import pandas as pd


# --- Q1〜Q5（選択式） ---
questions_data = [
    {
        "question": r"""
**Q1 講義で紹介したOSIの定義に照らして、オープンソースソフトウェア（OSS）の説明として最も適切なものはどれですか？**
(A)ソースコードを閲覧できれば、改変や再配布が禁止されていてもOSSである
(B)無料でダウンロードできるソフトウェアは、ソースコードが公開されていなくてもOSSである
(C)ライセンスの条件に従って、誰でも目的を問わず利用でき、ソースコードの改変や再配布ができる
(D)研究目的の利用のみを許可することが、OSSの必須条件である
(E)企業が開発を主導するソフトウェアは、公開条件にかかわらずOSSにはならない
""",
        "options": ["A", "B", "C", "D", "E"]
    },
    {
        "question": r"""
**Q2 doraの `dataflow.yml` で、`printer` ノードの入力を次のように記述しました。この記述の意味として最も適切なものはどれですか？**

```yaml
- id: printer
  path: printer.py
  inputs:
    message: doubler/doubled
```

(A)printer ノードが、doubler ノードのPython関数 doubled を直接呼び出す
(B)doubler ノードの doubled という出力を、printer ノードの message という入力へ接続する
(C)printer ノードの message という出力を、doubler ノードの入力へ接続する
(D)doubler/doubled という名前のデータファイルを、起動時に一度だけ読み込む
(E)printer ノードが受け取るすべての数値を、YAMLの設定だけで自動的に2倍にする
""",
        "options": ["A", "B", "C", "D", "E"]
    },
    {
        "question": r"""
**Q3 あるAIモデルでは、重みと推論コードを公開していますが、学習コードや学習データの出所・選別・前処理に関する情報は公開していません。講義で紹介したOpen WeightとOpen Source AIの違いに照らして、最も適切な説明はどれですか？**
(A)重みが公開されているため、それだけでOpen Source AIの要件を満たす
(B)推論コードが公開されていれば、学習コードやデータに関する情報を確認する必要はない
(C)手元で推論できるモデルは、利用条件にかかわらずOpen Source AIである
(D)論文にモデルの精度が記載されていれば、未公開の学習コードやデータに関する情報を補える
(E)重みの公開だけでは不十分であり、必要なコード、データに関する情報、公開・利用条件も確認する必要がある
""",
        "options": ["A", "B", "C", "D", "E"]
    },
    {
        "question": r"""
**Q4 OSSの不具合を見つけ、初めてIssueやPull Request（PR）で貢献するときの進め方として、最も適切なものはどれですか？**
(A)貢献ルールと既存Issueを確認し、環境・再現手順を示す。修正する場合は目的を絞り、動作確認の結果をPRに記載する
(B)不具合の再現手順はメンテナが調べるため、Issueには「動かない」とだけ記載する
(C)小さな修正は貢献として認められないため、関係のない機能追加も同じPRにまとめる
(D)AIが生成した修正であれば、投稿者による内容の理解や動作確認は不要である
(E)PRは提出した時点で作業完了となり、レビューで指摘を受けても修正する必要はない
""",
        "options": ["A", "B", "C", "D", "E"]
    },
    {
        "question": r"""
**Q5 オープンソースハードウェア（OSHW）の説明として、最も適切なものはどれですか？**
(A)完成品の写真を公開していれば、CADや回路図などの設計ファイルは非公開でもよい
(B)設計ファイルを公開する代わりに、その設計を使った製造や販売を禁止する仕組みである
(C)設計ファイルをダウンロードすれば、部品調達や加工・組立の費用は発生しない
(D)CAD・回路図・BOMなどの設計情報を公開し、ライセンスの条件に従った改変・製造・販売を可能にする
(E)ハードウェアと制御ソフトウェアには、必ず同一のライセンスを適用しなければならない
""",
        "options": ["A", "B", "C", "D", "E"]
    },
]

# Markdownの行末に半角スペース2つを自動付与（改行表示用）
for q in questions_data:
    q["question"] = '\n'.join([line.rstrip() + '  ' for line in q["question"].split('\n')])

# ウィジェット生成（Q1〜Q5）
widgets_list = []
for i, q in enumerate(questions_data):
    display(Markdown(q["question"]))
    rb = widgets.RadioButtons(
        options=q["options"],
        description="回答",
        style={'description_width': 'initial'}
    )
    widgets_list.append(rb)
    display(rb)

button_layout = widgets.Layout(width='200px', height='50px')
# --- 提出ボタン ---
submit_button = widgets.Button(description="✅ 回答をCSVに保存", layout=button_layout)
output = widgets.Output()

def on_submit_all(b):
    with output:
        clear_output()
        # Q1〜Q5の検証
        try:
            answers_mc = [rb.value for rb in widgets_list]  # 文字列で扱う
        except Exception:
            print("⚠️ Q1〜Q5に未回答があります。すべての選択問題に回答してください。")
            return
        if any(a is None for a in answers_mc):
            print("⚠️ Q1〜Q5に未回答があります。すべての選択問題に回答してください。")
            return

        # CSV生成：id=0..4 → Q1〜Q5
        labels = answers_mc
        df = pd.Series(labels, name="label")
        df.to_csv("submission.csv", index_label="id", encoding="utf-8")

        print("✅ 保存完了！ → 'submission.csv'")

submit_button.on_click(on_submit_all)
display(submit_button, output)
````
