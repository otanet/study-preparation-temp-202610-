```python
# このセルはそのまま実行してください（1セル完結）
import ipywidgets as widgets
from IPython.display import display, Markdown, clear_output
import pandas as pd


# --- Q1〜Q5（選択式） ---
questions_data = [
    {
        "question": r"""
**Q1 ROS 2 の CLI で，特定トピック /turtle1/pose のメッセージ内容をリアルタイムに表示する正しいコマンドはどれか．**
(1)ros2 topic list /turtle1/pose
(2)ros2 topic info /turtle1/pose
(3)ros2 topic echo /turtle1/pose
(4)ros2 topic pub /turtle1/pose geometry_msgs/msg/Pose
""",
        "options": ["1", "2", "3", "4"]
    },
    {
        "question": r"""
**Q2 すべてのトピックを録画し，各ファイルを zstd 形式で圧縮する rosbag2 コマンドとして最も適切なものはどれか．**
(1)ros2 bag record -a --compression-mode file --compression-format zstd
(2)ros2 bag record -a
(3)ros2 bag record -o subset
(4)ros2 bag record --lz4
""",
        "options": ["1", "2", "3", "4"]
    },
    {
        "question": r"""
**Q3 学習したポリシーをシミュレータにデプロイしたところ期待どおりに動作しなかった．学習が失敗しているのか，それとも汎化性能が足りていないのかを切り分けるために，講義で推奨された最初の手順はどれか．**
(1)本番環境のセンサノイズを除去するフィルタを追加する
(2)収集済みデータセットを使ってGround Truthと予測アクションを比較する
(3)学習率を10倍に設定して再学習する
(4)訓練データにフリップや回転などの画像前処理を追加する
""",
        "options": ["1", "2", "3", "4"]
    },
    {
        "question": r"""
**Q4 30 Hz のカメラ画像トピックと 100 Hz のジョイント状態トピックを同時に 3 分間 rosbag2 で録画した．このとき保存されるメッセージ数の合計として正しいものはどれか．**
(1)130 件
(2)5400 件
(3)18000 件
(4)23400 件
""",
        "options": ["1", "2", "3", "4"]
    },
    {
        "question": r"""
**Q5 講義で示された「Gazebo → rosbag2 → LeRobot での模倣学習」実践フローとして正しいものはどれか．**
(1)Gazebo でデータ収集 → Diffusion Policy 学習 → rosbag2 to LeRobot 変換 → デプロイ → 可視化
(2)rosbag2 で記録 → rosbag2 to LeRobot 変換 → データセット可視化 → ポリシー学習 → Gazebo へデプロイ
(3)データセット可視化 → rosbag2 to LeRobot 変換 → ポリシー学習 → デプロイ
(4)rosbag2 で記録 → 学習せずにそのまま Gazebo へデプロイ
""",
        "options": ["1", "2", "3", "4"]
    },
]

# Markdownの行末に半角スペース2つを自動付与（改行表示用）
for q in questions_data:
    q["question"] = '\n'.join([line.rstrip() + '  ' for line in q["question"].split('\n')])

# ウィジェット生成（Q1〜Q5）
widgets_list = []
for i, q in enumerate(questions_data):
    display(Markdown(q["question"]))
    rb = widgets.RadioButtons(
        options=q["options"],
        description="回答",
        style={'description_width': 'initial'}
    )
    widgets_list.append(rb)
    display(rb)

button_layout = widgets.Layout(width='200px', height='50px')
# --- 提出ボタン ---
submit_button = widgets.Button(description="✅ 回答をCSVに保存", layout=button_layout)
output = widgets.Output()

def on_submit_all(b):
    with output:
        clear_output()
        # Q1〜Q5の検証
        try:
            answers_mc = [rb.value for rb in widgets_list]  # 文字列で扱う
        except Exception:
            print("⚠️ Q1〜Q5に未回答があります。すべての選択問題に回答してください。")
            return
        if any(a is None for a in answers_mc):
            print("⚠️ Q1〜Q5に未回答があります。すべての選択問題に回答してください。")
            return

        # CSV生成：id=0..4 → Q1〜Q5
        labels = answers_mc
        df = pd.Series(labels, name="label")
        df.to_csv("submission.csv", index_label="id", encoding="utf-8")

        print("✅ 保存完了！ → 'submission.csv'")

submit_button.on_click(on_submit_all)
display(submit_button, output)
```
